# GPU相关操作

```python
import torch
import torch.nn as nn

# ---- GPU 基础操作 ----
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('使用设备:', device)
if torch.cuda.is_available():
    print('GPU 型号:', torch.cuda.get_device_name(0))
    print('显存总量:', torch.cuda.get_device_properties(0).total_memory / 1e9, 'GB')

# Tensor 移动到 GPU
x = torch.randn(3, 4)
x_gpu = x.to(device)       # 移到 device（GPU或CPU）
x_gpu2 = x.to(device)      # x.cuda() 直接移到GPU，无GPU时用 .to(device) 替代
x_back = x_gpu.cpu()       # 移回 CPU
print('\nTensor 设备:', x.device, '->', x_gpu.device)

# 模型移到 GPU
model = nn.Linear(4, 2)
model = model.to(device)
print('模型设备:', next(model.parameters()).device)

print()
print('GPU 最佳实践：')
print('  1. 训练开始前一次性把 model.to(device)')
print('  2. 每个 batch 把数据 .to(device)')
print('  3. 计算完后用 .item() 或 .cpu() 取回标量')
print('  4. 不要在循环中反复 .to()，有开销')
```
