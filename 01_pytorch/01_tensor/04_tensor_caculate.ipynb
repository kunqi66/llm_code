# 计算

```python
import torch
import numpy as np
```

## 1.基本四则

```python
# ── 1. 基本四则运算 ──────────────────────────────────────────────
t1 = torch.tensor([10, 4, 3, 23, 100], dtype=torch.float32)
print(t1)
print(t1 + 10)
print(t1 - 10)
print(t1 * 10)
print(t1 / 10)
print(t1 // 3)   # 整除
print(t1 % 3)    # 取余
print(torch.add(t1, 10))
print(torch.sub(t1, 10))
print(torch.mul(t1, 10))
print(torch.div(t1, 10))
```

## 2.幂运算与开方还有log

```python
print(t1 ** 2)
print(torch.pow(t1, 2))
print(torch.sqrt(t1))
print(torch.exp(t1[:3]))    # 取前三个，防止数值溢出
print(torch.log(t1))
```

## 3.矩阵运算

```python
a = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
b = torch.tensor([[5, 6], [7, 8]], dtype=torch.float32)
print(a @ b)               # @ 运算符
print(torch.mm(a, b))      # 二维矩阵专用
print(torch.matmul(a, b))  # 通用（支持广播）

# 元素级乘法（Hadamard 积）
print(a * b)
print(torch.mul(a, b))
```

## 4.聚合运算

```python
print(t1.sum())
print(t1.mean())
print(t1.max())
print(t1.min())
print(t1.prod())           # 连乘
print(t1.std())            # 标准差
```

## 5. 比较运算

```python
print(t1 > 10)
print(t1 == 10)
print(t1 >= 10)
print(torch.eq(t1, 10))
```
