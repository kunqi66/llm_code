```python
import torch
import torch.nn as nn
```

## 1.常量列表直接创建


```python
tensor1 = torch.tensor(10)
print(tensor1.dim())
print(tensor1)
```

```python
tensor2 = torch.tensor([10,20])
print(tensor2.dim())
print(tensor2)
```

```python
tensor3 = torch.tensor([[10],[20]])
print(tensor3.dim())
print(tensor3)
```

## 2.随机创建


```python
print("正态分布创建", torch.randn(2,3))

print(torch.normal(3,5,[3,4,5]))
```

```python
print("均匀随机", torch.rand(2,3,4))
```

## 3. ndarray创建

```python
import numpy as np
```

```python
print(torch.tensor(np.array([2,3,4,4])))
```

使用ndarray创建的张量与ndarray共享内存

```python
# 与numpy共享内存
np_num = np.array([[2,3,4],[3,4,5]])

torch_num = torch.from_numpy(np_num)

print(torch_num)
torch_num[0][1] = 0
print(np_num)
print(torch_num)
```

## 4.特殊张量的创建

```python
print('全0张量',torch.zeros(2,3))
print('模仿全0',torch.zeros_like(torch_num))
print('等差数列',torch.arange(1,100,3))
print('线性等分',torch.linspace(1,5,3))
```
