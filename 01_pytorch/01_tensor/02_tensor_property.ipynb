# 张量属性

```python
import torch
import torch.nn as nn
```

```python
tensor = torch.normal(3,5,[3,4,5])
```

```python
print('tensor维度',tensor.ndim)
print('tensor形状',tensor.shape)
print('tensor元素数量',tensor.numel())
print('tensor运行设备',tensor.device)
print('tensor数据类型',tensor.dtype)
```

```python
# 常见数据类型
print('\n--- 数据类型 ---')
print('float32:', torch.ones(3).dtype)           # 默认浮点类型
print('int64:  ', torch.ones(3, dtype=torch.long).dtype)
print('bool:   ', torch.ones(3).bool().dtype)
```
