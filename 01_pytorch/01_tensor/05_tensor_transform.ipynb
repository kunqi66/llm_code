## 转化

```python
import numpy as np
import torch
```

## 1.数据类型转换

```python
t1 = torch.tensor([2,3,4])
print(t1.dtype)
```

```python
t1 = t1.to(torch.float32)
print(t1.dtype)
```

```python
t1 = t1.type(torch.int32)
print(t1.dtype)
```

```python
# 便捷转换
t1 = t1.float()             # → float32
t1 = t1.double()            # → float64
t1 = t1.int()               # → int32
t1 = t1.long()              # → int64
t1 = t1.bool()              # → bool
print(t1.dtype)
```

## 2.设备转化

```python
t2 = torch.tensor([1.0,2.0,3.0])
print(t2)
print(t2.device)
```

```python
if torch.cuda.is_available():
    t2_gpu = t2.to('cuda')
    print(t2_gpu.device)
    t2_back = t2_gpu.cpu()
    print(t2_back.device)
    t2 = t2.cuda()
    print(t2.device)
```
