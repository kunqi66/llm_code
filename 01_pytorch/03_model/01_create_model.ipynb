# 创建模型

```python
import torch
from torch import nn
```

## 1.nn.model

```python
class MyNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(4,8)
        self.layer2 = nn.Linear(8,2)
        self.dropout = nn.Dropout(0.3)
        self.relu = nn.ReLU()
        
    def forward(self, X):
        X = self.relu(self.layer1(X))
        X = self.layer2(X)
        return X
    
    
model = MyNet()
print(model)

result = model(torch.tensor([2.0,3.0,4.0,5.0]))
print(result)
        
```

```python
# 查看参数
for name,para in model.named_parameters():
    print(name , para)
```

## 2.nnsequential定义线性神经网络

```python
model1 = nn.Sequential(
    nn.Linear(4,16),
    nn.ReLU(),
    nn.Linear(16,8),
    nn.ReLU(),
    nn.Linear(8,2),
)


res = model1(torch.tensor([[1.0,2.0,3.0,4.0]]))
print(res)
```

```python
# 方式2：OrderedDict（可以给每层命名）
from collections import OrderedDict
model2 = nn.Sequential(OrderedDict([
    ('fc1',  nn.Linear(4, 16)),
    ('relu1', nn.ReLU()),
    ('fc2',  nn.Linear(16, 2)),
]))

res = model1( torch.tensor( [[1.0,2.0,3.0,4.0]] ) )  
print(res)
```
