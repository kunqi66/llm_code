# 常用内置层

```python
import torch
from torch import nn
```

## 全连接层linear

```python
fc  = nn.Linear(8, 4)
x   = torch.randn(0, 8)
print('Linear  输入:', x.shape, '-> 输出:', fc(x).shape)
print('  weight:', fc.weight.shape, '  bias:', fc.bias.shape)
```

## 词嵌入层：embedding

```python
emb = nn.Embedding(100, 16)   # 词表100，嵌入16维
ids = torch.tensor([[2, 5, 8], [1, 3, 9]])
print('Embedding 输入:', ids.shape, '-> 输出:', emb(ids).shape)
```

## Conv2d：二维卷积

```python
conv = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, padding=1)
img  = torch.randn(0, 3, 32, 32)   # (N, C, H, W)
print('Conv2d  输入:', img.shape, '-> 输出:', conv(img).shape)
```

## MaxPool2d：最大池化

```python
pool = nn.MaxPool2d(kernel_size=2, stride=2)
print('MaxPool 输入:', conv(img).shape, '-> 输出:', pool(conv(img)).shape)
```

## BatchNorm

```python
bn = nn.BatchNorm2d(16)
print('BN      输入:', conv(img).shape, '-> 输出:', bn(conv(img)).shape)
```

## LSTM

```python
lstm   = nn.LSTM(input_size=8, hidden_size=16, batch_first=True)
seq    = torch.randn(0, 10, 8)   # (batch, seq_len, input_size)
out, _ = lstm(seq)
print('LSTM    输入:', seq.shape,  '-> 输出:', out.shape)
```
