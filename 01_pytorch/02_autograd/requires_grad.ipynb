# 自动求导

```python
import torch
import numpy as np
```

## 1.requirs_grad

```python
x = torch.tensor([2.0, 3.0], requires_grad=True)
print("x", x)
```

```python
y = x ** 2
z = y.sum()
print(y)
print(z)
```

```python
z.backward()
```

```python
print(x.grad)
print("验证: 2 * x = ",(2 * x).tolist())
```

## 2.多层计算

```python
x = torch.tensor(3.0 , requires_grad = True)
y = 2 * x **3 + x**2 +8 * x +7
z = y.sum()
z.backward()
print(x.grad)
```

```python
# ---- torch.no_grad()：推理时不需要梯度，节省内存和计算 ----
with torch.no_grad():
    y = x * 2
    print(y.requires_grad)
```

```python
# ---- detach()：从计算图中分离 ----
x = torch.tensor([1.0, 2.0], requires_grad=True)
y = x * 2
y_detached = y.detach()
print('detach 后 requires_grad:', y_detached.requires_grad)  # False
```

## 3.梯度累计问题

```python
# ---- 梯度累积问题 ----
import torch

x = torch.tensor(2.0, requires_grad=True)

for i in range(3):
    y = x ** 2
    y.backward()
    print(f'第{i+1}次反向传播后 x.grad = {x.grad.item()}')
    # 注意：梯度会累加！不清零的话第2次是8，第3次是12

print('\n--- 使用 zero_grad() 修复 ---')
x.grad = None   # 或 x.grad.zero_()
for i in range(3):
    y = x ** 2
    y.backward()
    print(f'第{i+1}次 x.grad = {x.grad.item()}')
    x.grad.zero_()  # 每次反向传播前清零
```
